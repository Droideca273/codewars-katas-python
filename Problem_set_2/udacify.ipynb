```python
def countdown(n):
    while n>0:
            print(n)
            n=n-1
    print("Blastoff!")        
countdown(3)
```

```python
def print_multiplication_table(n):
    for i in range(1,n+1):
        for j in range(1,n+1):
            print(str(i)+" * "+str(j)+" = "+str(i*j))
print_multiplication_table(3)
```

```python
def euros(n):
    mon5=0
    mon2=0
    mon1=0
    if n/5>=1:
        mon5=int(n/5)
        n=n-(mon5*5)
    if n/2>=1:
         mon2=int(n/2)
         n=n-(mon2*2)
    if n/1>=1:
        mon1=n
    return(mon5, mon2, mon1)
print(euros(27))
```
