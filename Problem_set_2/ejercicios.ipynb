# Udacify

```python
def udacify(input):
    return "U" + input


print(udacify("dacians"))
```

# Median

```python
def median(num1, num2, num3):
    return (num1 + num2 + num3) / 3


median(9, 3, 6)
```

# Blastoff!

```python
def countdown(n):
    while n > 0:
        print(n)
        n = n - 1
    print("Blastoff!")


countdown(3)
```

# Find Last

```python
def find_last(string, target):
    return string.find(target, -1)


print(find_last("Caracola", "a"))
```

# Multiplication Table

```python
def print_multiplication_table(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            print(str(i) + " * " + str(j) + " = " + str(i * j))


print_multiplication_table(3)
```

# Monedas

```python
def euros(n):
    monedas=[5,2,1]
    cantidad_monedas=[0,0,0]
    for i in range(len(monedas)):
        if n//monedas[i]>=1:
            cantidad_monedas[i]=n//monedas[i]
            n=n%monedas[i]
    return (tuple(cantidad_monedas))
euros(67)
```
